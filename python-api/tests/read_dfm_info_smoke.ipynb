# Read DFM Info Smoke Test

This notebook uses the public ArcRho Python API. `ArcRhoClient()` reuses the ArcRho frontend Server Connection from `%APPDATA%\Arco\workspace_paths.json`, so no server-root path is needed here.

```python
from arcrho_api import ArcRhoClient

PROJECT_NAME = "NJ_Annual_Prod_202605_Fake"
RESERVING_CLASS = r"HPPREF\HO+DF\NJ\Legacy\HOL"
DFM_NAME = "C 12 - CWP DFM w/ Selected LDFs"

client = ArcRhoClient(read_only=True)
project = client.project(PROJECT_NAME)
rc = project.reserving_class(RESERVING_CLASS)
dfm_c12 = rc.dfm(DFM_NAME)

dfm_c12.info()
```
